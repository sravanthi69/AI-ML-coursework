```python
#remove unnecessary columns
df = df.drop(columns=["car_ID", "CarName"])
```

```python
#seperating features and target into separate datasets
x=df.drop(columns=['price'])
y=df['price']
```

```python
print("Feature data shape:",x.shape)
print("Target data shape:", y.shape)
```

```python
print("Numerical columns:")
print(x.select_dtypes(include=["int64", "float64"]).columns)

print("\nCategorical columns:")
print(x.select_dtypes(include=["object"]).columns)
```

```python
#convert categorical features into numerical ones using one-hot encoding
x_encoded = pd.get_dummies(x, drop_first=True)
```

```python
print("Original shape:", x.shape)
print("Encoded shape:", x_encoded.shape)
```

```python
x_encoded.head()
```

SPLIT THE DATA INTO TRAIN AND TEST

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_encoded,y,test_size=0.20,random_state=42)
```

```python
print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
```

```python
linear_model = LinearRegression()

ridge_model = Ridge(alpha=1.0)

lasso_model = Lasso(alpha=1.0, max_iter=10000)
```

```python
#train the model
linear_model.fit(x_train_scaled, y_train)

ridge_model.fit(x_train_scaled, y_train)

lasso_model.fit(x_train_scaled, y_train)
```

```python
print("Linear Regression model trained")
print("Ridge Regression model trained")
print("Lasso Regression model trained")
```

```python
linear_pred = linear_model.predict(x_test_scaled)
ridge_pred = ridge_model.predict(x_test_scaled)
lasso_pred = lasso_model.predict(x_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
def evaluate_model(model_name, actual, predicted):

    mae = mean_absolute_error(actual, predicted)
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(actual, predicted)

    return {"Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2 }
```

```python
linear_results = evaluate_model("Linear Regression", y_test, linear_pred)

ridge_results = evaluate_model( "Ridge Regression", y_test, ridge_pred)

lasso_results = evaluate_model("Lasso Regression", y_test, lasso_pred)
```

```python
results_df = pd.DataFrame([
    linear_results,
    ridge_results,
    lasso_results
])

results_df
```

LINEAR REGRESSION PERFORMED BEST. Its because it had the lowest error values and the highest R² score of about 0.89. Lasso gave very similar results, while Ridge performed slightly worse. Ridge reduced the coefficients, and Lasso could reduce some coefficients to zero. For this dataset, regularization did not improve the model, so Linear Regression generalized best.
